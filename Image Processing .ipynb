```python
from PIL import Image
import matplotlib.pyplot as plt

image = Image.open("my_object.png")

plt.imshow(image)
plt.axis("off")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
gray = cv2.cvtColor(np.array(image), cv2.COLOR_RGB2GRAY)

plt.imshow(gray, cmap="gray")
plt.axis("off")
plt.show()
```

```python
edges = cv2.Canny(gray, 100, 200)

plt.imshow(edges, cmap="gray")
plt.axis("off")
plt.show()
```

```python
contours, _ = cv2.findContours(
    edges,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

print("Number of boundaries detected:", len(contours))
```

```python
# Select the detected boundary
contour = contours[0]

# Find the rectangular boundary around the object
x, y, width, height = cv2.boundingRect(contour)

print("Width  :", width, "cm")
print("Height :", height, "cm")
```

```python
output = np.array(image).copy()

cv2.rectangle(
    output,
    (x, y),
    (x + width, y + height),
    (0, 255, 0),
    3
)

plt.figure(figsize=(8, 8))
plt.imshow(output)
plt.axis("off")
plt.show()
```
